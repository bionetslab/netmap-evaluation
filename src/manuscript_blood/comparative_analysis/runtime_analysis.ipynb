```python
import pandas as pd
```

```python
runtimes = pd.read_csv('/data_nfs/og86asub/netmap/netmap-evaluation/netmap/case_studies/blood/time_tracker_repeat.tsv', index_col=0)
```

```python
runtimes.columns = ['experiment', 'time']
```

```python
runtimes['time_min'] = runtimes['time']/60
```

```python
runtimes['experiment'] = runtimes['experiment'].str.split('-').str[:-1].str.join('-')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Combine the dataframes

# Set the figure size for better readability
plt.figure(figsize=(1.8, 2))
# Group by 'category' on the x-axis
sns.boxplot(data=runtimes, x='experiment', y='time_min', palette='Set2', legend=False)
plt.title('Runtimes for 10 repeats')
plt.xticks(rotation=45, ha='right')
plt.ylabel('runtime [minutes]')
plt.xlabel('')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig
plt.show()
```
